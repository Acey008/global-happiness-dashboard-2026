```python


# =============================================================================
# GLOBAL HAPPINESS AND QUALITY OF LIFE ANALYSIS | 2026
# Exploratory Data Analysis and Statistical Assessment
# =============================================================================

# 1. LIBRARIES AND CONFIGURATION

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

from IPython.display import display
from google.colab import files

sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams.update({
    "figure.figsize": (11, 6),
    "figure.dpi": 120,
    "axes.titlesize": 14,
    "axes.labelsize": 11
})

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:.3f}".format)


# =============================================================================
# 2. DATA VALIDATION AND QUALITY ASSESSMENT
# =============================================================================

data = df.copy()

data.columns = (
    data.columns.str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

data["country"] = data["country"].astype(str).str.strip()

print("DATASET OVERVIEW")
print(f"Observations: {data.shape[0]}")
print(f"Variables: {data.shape[1]}")
print(f"Unique countries: {data['country'].nunique()}")
print(f"Missing values: {data.isna().sum().sum()}")
print(f"Duplicate records: {data.duplicated().sum()}")

print("\nVARIABLE TYPES")
display(data.dtypes.to_frame("Data Type"))

print("\nMISSING VALUES BY VARIABLE")
display(data.isna().sum().to_frame("Missing Values"))

print("\nREPEATED COUNTRY NAMES")
repeated_countries = data[
    data.duplicated(subset="country", keep=False)
].sort_values(["country", "rank"])

display(repeated_countries)

print("\nRANK VALIDATION")
print("Unique ranks:", data["rank"].nunique())
print("Minimum rank:", data["rank"].min())
print("Maximum rank:", data["rank"].max())

rank_consistent = (
    data.sort_values("rank")["score"]
    .is_monotonic_decreasing
)
print("Scores follow descending rank:", rank_consistent)

# Identify inconsistent observations without deleting source records.
country_conflicts = (
    data.groupby("country")
    .agg(
        records=("rank", "size"),
        distinct_scores=("score", "nunique")
    )
    .query("records > 1")
)

print("\nCOUNTRY-LEVEL CONSISTENCY")
display(country_conflicts)

# Retain the complete source dataset.
# Country duplicates require verification before deduplication.

analysis_df = data.copy()


# =============================================================================
# 3. DESCRIPTIVE STATISTICAL ANALYSIS
# =============================================================================

indicators = [
    "gdp_per_capita",
    "social_support",
    "healthy_life_expectancy",
    "freedom",
    "generosity",
    "corruption"
]

print("\nDESCRIPTIVE STATISTICS")
display(
    analysis_df[indicators + ["score"]]
    .describe()
    .T
    .round(3)
)

summary = pd.DataFrame({
    "Metric": [
        "Mean happiness score",
        "Median happiness score",
        "Standard deviation",
        "Minimum happiness score",
        "Maximum happiness score",
        "Interquartile range"
    ],
    "Value": [
        analysis_df["score"].mean(),
        analysis_df["score"].median(),
        analysis_df["score"].std(),
        analysis_df["score"].min(),
        analysis_df["score"].max(),
        analysis_df["score"].quantile(0.75)
        - analysis_df["score"].quantile(0.25)
    ]
})

print("\nHAPPINESS SCORE SUMMARY")
display(summary.round(3))


# =============================================================================
# 4. COUNTRY RANKINGS
# =============================================================================

top_10 = analysis_df.nlargest(10, "score")
bottom_10 = analysis_df.nsmallest(10, "score")

print("\nTOP 10 HIGHEST-RANKED COUNTRIES")
display(top_10[["rank", "country", "score"]])

print("\nBOTTOM 10 LOWEST-RANKED COUNTRIES")
display(bottom_10[["rank", "country", "score"]])

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

sns.barplot(
    data=top_10,
    x="score",
    y="country",
    color="#267D79",
    ax=axes[0]
)

axes[0].set_title("Top 10 Countries by Happiness Score")
axes[0].set_xlabel("Happiness Score")
axes[0].set_ylabel("")

sns.barplot(
    data=bottom_10.sort_values("score"),
    x="score",
    y="country",
    color="#C77C62",
    ax=axes[1]
)

axes[1].set_title("Bottom 10 Countries by Happiness Score")
axes[1].set_xlabel("Happiness Score")
axes[1].set_ylabel("")

plt.tight_layout()
plt.show()


# =============================================================================
# 5. DISTRIBUTION OF HAPPINESS SCORES
# =============================================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(
    analysis_df["score"],
    bins=15,
    kde=True,
    color="#438C91",
    ax=axes[0]
)

axes[0].set_title("Distribution of Happiness Scores")
axes[0].set_xlabel("Happiness Score")
axes[0].set_ylabel("Frequency")

sns.boxplot(
    x=analysis_df["score"],
    color="#8FB9B2",
    ax=axes[1]
)

axes[1].set_title("Happiness Score: Spread and Outliers")
axes[1].set_xlabel("Happiness Score")

plt.tight_layout()
plt.show()

print("\nDISTRIBUTION CHARACTERISTICS")
print("Skewness:", round(analysis_df["score"].skew(), 3))
print("Kurtosis:", round(analysis_df["score"].kurt(), 3))


# =============================================================================
# 6. CORRELATION ANALYSIS
# =============================================================================

correlation_matrix = (
    analysis_df[indicators + ["score"]]
    .corr(method="pearson")
)

print("\nPEARSON CORRELATION MATRIX")
display(correlation_matrix.round(3))

plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="BrBG",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title("Correlation Between Happiness and Quality of Life Indicators")
plt.tight_layout()
plt.show()

score_correlations = (
    correlation_matrix["score"]
    .drop("score")
    .sort_values(ascending=False)
)

print("\nCORRELATION WITH HAPPINESS SCORE")
display(
    score_correlations.to_frame(
        "Pearson Correlation"
    ).round(3)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=score_correlations.values,
    y=score_correlations.index,
    color="#438C91"
)

plt.axvline(0, color="black", linewidth=0.8)
plt.title("Association of Quality of Life Indicators with Happiness")
plt.xlabel("Pearson Correlation Coefficient")
plt.ylabel("")
plt.xlim(-1, 1)
plt.tight_layout()
plt.show()


# =============================================================================
# 7. BIVARIATE RELATIONSHIPS
# =============================================================================

fig, axes = plt.subplots(2, 3, figsize=(17, 10))

for ax, indicator in zip(axes.flat, indicators):
    sns.regplot(
        data=analysis_df,
        x=indicator,
        y="score",
        ax=ax,
        scatter_kws={"alpha": 0.65, "s": 35},
        line_kws={"color": "#B45F55"}
    )

    ax.set_title(
        indicator.replace("_", " ").title()
        + " vs Happiness"
    )
    ax.set_xlabel(indicator.replace("_", " ").title())
    ax.set_ylabel("Happiness Score")

plt.suptitle(
    "Relationships Between Happiness and Quality of Life Indicators",
    fontsize=16,
    y=1.01
)

plt.tight_layout()
plt.show()


# =============================================================================
# 8. INTERACTIVE GLOBAL HAPPINESS MAP
# =============================================================================

fig = px.choropleth(
    analysis_df,
    locations="country",
    locationmode="country names",
    color="score",
    hover_name="country",
    hover_data={
        "rank": True,
        "score": ":.3f",
        "country": False
    },
    color_continuous_scale="Tealgrn",
    title="Global Distribution of Happiness Scores | 2026",
    labels={"score": "Happiness Score"}
)

fig.update_layout(
    geo=dict(
        showframe=False,
        showcoastlines=True,
        projection_type="natural earth"
    ),
    height=600
)

fig.show()

# Country-name matching and repeated country labels
# must be verified before publication.


# =============================================================================
# 9. INTERACTIVE COUNTRY COMPARISON
# =============================================================================

selected_countries = [
    "Finland",
    "Denmark",
    "Costa Rica",
    "India",
    "United Kingdom",
    "United States"
]

comparison_df = analysis_df[
    analysis_df["country"].isin(selected_countries)
].copy()

print("\nSELECTED COUNTRY COMPARISON")
display(
    comparison_df[
        ["country", "rank", "score"] + indicators
    ].sort_values("rank")
)

fig = px.bar(
    comparison_df,
    x="country",
    y="score",
    color="score",
    color_continuous_scale="Tealgrn",
    title="Happiness Score Comparison: Selected Countries",
    labels={
        "country": "Country",
        "score": "Happiness Score"
    },
    hover_data=["rank"]
)

fig.update_layout(
    height=500,
    xaxis_tickangle=-30
)

fig.show()


# =============================================================================
# 10. INDICATOR PROFILE COMPARISON
# =============================================================================

# Standardisation allows indicators with different scales
# to be compared without treating raw values as equivalent.

profile_df = analysis_df[
    analysis_df["country"].isin(selected_countries)
].copy()

indicator_std = analysis_df[indicators].std()

valid_indicators = indicator_std[
    indicator_std > 0
].index.tolist()

standardised = (
    profile_df[valid_indicators]
    - analysis_df[valid_indicators].mean()
) / indicator_std[valid_indicators]

standardised.index = profile_df["country"].values

print("\nSTANDARDISED INDICATOR PROFILES")
display(standardised.round(2))

if not standardised.empty:
    plt.figure(figsize=(12, 6))

    sns.heatmap(
        standardised,
        annot=True,
        fmt=".2f",
        cmap="RdBu",
        center=0,
        linewidths=0.5
    )

    plt.title(
        "Selected Countries: Standardised Quality of Life Indicators"
    )
    plt.xlabel("Indicator")
    plt.ylabel("Country")
    plt.tight_layout()
    plt.show()


# =============================================================================
# 11. HAPPINESS SCORE VARIABILITY
# =============================================================================

q1 = analysis_df["score"].quantile(0.25)
q3 = analysis_df["score"].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = analysis_df[
    (analysis_df["score"] < lower_bound)
    | (analysis_df["score"] > upper_bound)
]

print("\nIQR-BASED OUTLIER ASSESSMENT")
print("Lower bound:", round(lower_bound, 3))
print("Upper bound:", round(upper_bound, 3))
print("Observations outside bounds:", len(outliers))

display(
    outliers[["rank", "country", "score"]]
    .sort_values("score")
)

# Outliers are retained because extreme country scores
# may represent meaningful observations.


# =============================================================================
# 12. EXPORT ANALYTICAL OUTPUTS
# =============================================================================

analysis_df.to_csv(
    "global_happiness_2026_analysis.csv",
    index=False
)

correlation_matrix.to_csv(
    "happiness_correlation_matrix.csv"
)

summary.to_csv(
    "happiness_summary_statistics.csv",
    index=False
)

top_10.to_csv(
    "top_10_happiness_countries.csv",
    index=False
)

bottom_10.to_csv(
    "bottom_10_happiness_countries.csv",
    index=False
)

print("\nEXPORTED ANALYTICAL FILES")
print("global_happiness_2026_analysis.csv")
print("happiness_correlation_matrix.csv")
print("happiness_summary_statistics.csv")
print("top_10_happiness_countries.csv")
print("bottom_10_happiness_countries.csv")


# =============================================================================
# 13. ANALYTICAL LIMITATIONS
# =============================================================================

limitations = [
    "The dataset represents a single reporting year.",
    "Cross-sectional correlations do not establish causality.",
    "Repeated country labels require source verification.",
    "Indicators must be interpreted according to their source definitions.",
    "Country-level observations do not represent individual experiences.",
    "Geographical name matching may require standardisation."
]

print("\nANALYTICAL LIMITATIONS")

for item in limitations:
    print("-", item)
```

```python

# =============================================================================
# COUNTRY-LEVEL DATA QUALITY VALIDATION
# =============================================================================

# Identify countries appearing more than once
country_counts = (
    analysis_df["country"]
    .value_counts()
    .reset_index()
)

country_counts.columns = ["country", "record_count"]

repeated_countries = country_counts[
    country_counts["record_count"] > 1
]

print("REPEATED COUNTRY NAMES")
display(repeated_countries)

# Examine the corresponding records
duplicate_country_records = analysis_df[
    analysis_df["country"].isin(repeated_countries["country"])
].sort_values(["country", "rank"])

print("\nRECORDS REQUIRING VERIFICATION")
display(
    duplicate_country_records[
        ["rank", "country", "score"]
    ]
)

# Assess differences between repeated records
country_variation = (
    duplicate_country_records
    .groupby("country")
    .agg(
        records=("rank", "count"),
        minimum_score=("score", "min"),
        maximum_score=("score", "max")
    )
)

country_variation["score_difference"] = (
    country_variation["maximum_score"]
    - country_variation["minimum_score"]
)

print("\nCOUNTRY SCORE CONSISTENCY")
display(country_variation.round(3))
```

```python

# =============================================================================
# INVESTIGATION OF CONFLICTING COUNTRY RECORDS
# =============================================================================

# Compare all variables for countries with repeated observations.

conflicting_records = (
    analysis_df[
        analysis_df["country"].duplicated(keep=False)
    ]
    .sort_values(["country", "rank"])
    .copy()
)

comparison_columns = [
    "country",
    "rank",
    "score",
    "gdp_per_capita",
    "social_support",
    "healthy_life_expectancy",
    "freedom",
    "generosity",
    "corruption"
]

print("CONFLICTING COUNTRY OBSERVATIONS")
display(conflicting_records[comparison_columns])

# Compare the position of each repeated observation.

rank_comparison = (
    conflicting_records
    .groupby("country")
    .agg(
        highest_rank=("rank", "min"),
        lowest_rank=("rank", "max"),
        highest_score=("score", "max"),
        lowest_score=("score", "min")
    )
)

rank_comparison["rank_difference"] = (
    rank_comparison["lowest_rank"]
    - rank_comparison["highest_rank"]
)

rank_comparison["score_difference"] = (
    rank_comparison["highest_score"]
    - rank_comparison["lowest_score"]
)

print("\nRANK AND SCORE DIFFERENCES")
display(rank_comparison.round(3))

# Identify whether repeated records occur in distinct rank ranges.

print("\nDISTRIBUTION OF CONFLICTING RECORDS BY RANK")
display(
    conflicting_records[
        ["country", "rank", "score"]
    ].sort_values("rank")
)
```

```python

# =============================================================================
# SOURCE DATA VALIDATION
# =============================================================================

# Check whether the existing rankings form a complete sequence.

expected_ranks = set(range(1, len(analysis_df) + 1))
observed_ranks = set(analysis_df["rank"])

print("Missing rank numbers:")
print(sorted(expected_ranks - observed_ranks))

print("\nRepeated rank numbers:")
print(
    analysis_df.loc[
        analysis_df["rank"].duplicated(keep=False),
        ["country", "rank", "score"]
    ]
)

# Examine the ordering of scores across the full dataset.

ranked_data = analysis_df.sort_values("rank").copy()

ranked_data["previous_score"] = ranked_data["score"].shift(1)

ranked_data["score_order_violation"] = (
    ranked_data["score"] > ranked_data["previous_score"]
)

print("\nRank-score inconsistencies:")
display(
    ranked_data.loc[
        ranked_data["score_order_violation"],
        ["rank", "country", "score", "previous_score"]
    ]
)
```

```python

# =============================================================================
# GLOBAL HAPPINESS & QUALITY OF LIFE DASHBOARD | 2026
# Streamlit Application Development and GitHub Preparation
# =============================================================================

from pathlib import Path
import pandas as pd
from google.colab import files
import zipfile

project = Path("global_happiness_dashboard")
project.mkdir(exist_ok=True)

# 1. PREPARE THE DASHBOARD DATASET

source_data = df.copy()
source_data["country"] = source_data["country"].str.strip()

duplicate_mask = source_data["country"].duplicated(keep=False)

excluded = source_data.loc[duplicate_mask].copy()
clean_data = source_data.loc[~duplicate_mask].copy()

clean_data = clean_data.sort_values(
    "score", ascending=False
).reset_index(drop=True)

clean_data["analysis_rank"] = range(1, len(clean_data) + 1)

clean_data.to_csv(
    project / "happiness_2026_clean.csv",
    index=False
)

excluded.to_csv(
    project / "excluded_country_records.csv",
    index=False
)

print("DATA PREPARATION SUMMARY")
print("Original records:", len(source_data))
print("Countries included:", clean_data["country"].nunique())
print("Countries excluded:", excluded["country"].nunique())


# =============================================================================
# 2. STREAMLIT DASHBOARD APPLICATION
# =============================================================================

app_code = r'''
import streamlit as st
import pandas as pd
import plotly.express as px

st.set_page_config(
    page_title="Global Happiness Dashboard 2026",
    page_icon="🌍",
    layout="wide"
)

@st.cache_data
def load_data():
    return pd.read_csv("happiness_2026_clean.csv")

df = load_data()

indicators = [
    "gdp_per_capita",
    "social_support",
    "healthy_life_expectancy",
    "freedom",
    "generosity",
    "corruption"
]

st.title("Global Happiness & Quality of Life Dashboard")
st.caption("2026 | Cross-Country Exploratory Analysis")

st.info(
    "Data-quality note: 17 countries with conflicting "
    "duplicate observations were excluded. "
    "Results cover 113 countries, not the full original "
    "147-record dataset. Rankings are recalculated "
    "within the retained sample."
)

st.sidebar.header("Dashboard Filters")

selected_countries = st.sidebar.multiselect(
    "Select countries",
    sorted(df["country"].unique()),
    default=sorted(df["country"].unique())
)

filtered = df[df["country"].isin(selected_countries)]

if filtered.empty:
    st.warning("Select at least one country.")
    st.stop()

# -------------------------------------------------------------------------
# OVERVIEW METRICS
# -------------------------------------------------------------------------

col1, col2, col3, col4 = st.columns(4)

col1.metric("Countries", len(filtered))
col2.metric("Average Happiness", f"{filtered['score'].mean():.2f}")
col3.metric("Highest Score", f"{filtered['score'].max():.2f}")
col4.metric("Lowest Score", f"{filtered['score'].min():.2f}")

# -------------------------------------------------------------------------
# INTERACTIVE WORLD MAP
# -------------------------------------------------------------------------

st.subheader("Global Happiness Distribution")

fig_map = px.choropleth(
    filtered,
    locations="country",
    locationmode="country names",
    color="score",
    hover_name="country",
    hover_data=["analysis_rank"],
    color_continuous_scale="Tealgrn",
    projection="natural earth",
    labels={"score": "Happiness Score"}
)

fig_map.update_layout(
    height=550,
    margin=dict(l=0, r=0, t=10, b=0)
)

st.plotly_chart(fig_map, use_container_width=True)

# -------------------------------------------------------------------------
# COUNTRY RANKINGS
# -------------------------------------------------------------------------

st.subheader("Country Rankings")

left, right = st.columns(2)

top_countries = filtered.nlargest(10, "score")
bottom_countries = filtered.nsmallest(10, "score")

with left:
    fig_top = px.bar(
        top_countries.sort_values("score"),
        x="score",
        y="country",
        orientation="h",
        title="Top 10 Countries",
        color="score",
        color_continuous_scale="Tealgrn"
    )
    st.plotly_chart(fig_top, use_container_width=True)

with right:
    fig_bottom = px.bar(
        bottom_countries.sort_values("score"),
        x="score",
        y="country",
        orientation="h",
        title="Bottom 10 Countries",
        color="score",
        color_continuous_scale="OrRd"
    )
    st.plotly_chart(fig_bottom, use_container_width=True)

# -------------------------------------------------------------------------
# HAPPINESS SCORE DISTRIBUTION
# -------------------------------------------------------------------------

st.subheader("Distribution of Happiness Scores")

fig_hist = px.histogram(
    filtered,
    x="score",
    nbins=15,
    marginal="box",
    title="Happiness Score Distribution",
    labels={"score": "Happiness Score"}
)

st.plotly_chart(fig_hist, use_container_width=True)

# -------------------------------------------------------------------------
# CORRELATION ANALYSIS
# -------------------------------------------------------------------------

st.subheader("Correlation Analysis")

corr = filtered[indicators + ["score"]].corr()

fig_corr = px.imshow(
    corr,
    text_auto=".2f",
    color_continuous_scale="RdBu",
    zmin=-1,
    zmax=1,
    title="Pearson Correlation Matrix"
)

st.plotly_chart(fig_corr, use_container_width=True)

indicator = st.selectbox(
    "Select an indicator to examine",
    indicators
)

fig_scatter = px.scatter(
    filtered,
    x=indicator,
    y="score",
    hover_name="country",
    title=f"{indicator.replace('_', ' ').title()} vs Happiness Score",
    labels={"score": "Happiness Score"}
)

st.plotly_chart(fig_scatter, use_container_width=True)

# -------------------------------------------------------------------------
# COUNTRY COMPARISON
# -------------------------------------------------------------------------

st.subheader("Country Comparison")

comparison = st.multiselect(
    "Select countries to compare",
    sorted(df["country"].unique()),
    default=[
        country for country in
        ["Finland", "Denmark", "India", "United Kingdom"]
        if country in df["country"].values
    ]
)

comparison_data = df[df["country"].isin(comparison)]

if not comparison_data.empty:
    fig_compare = px.bar(
        comparison_data,
        x="country",
        y="score",
        color="country",
        title="Happiness Score by Country"
    )
    st.plotly_chart(fig_compare, use_container_width=True)

    st.dataframe(
        comparison_data[
            ["country", "analysis_rank", "score"] + indicators
        ],
        use_container_width=True,
        hide_index=True
    )

# -------------------------------------------------------------------------
# DATA EXPLORER AND DOWNLOAD
# -------------------------------------------------------------------------

st.subheader("Explore the Dataset")

st.dataframe(
    filtered,
    use_container_width=True,
    hide_index=True
)

st.download_button(
    label="Download Filtered Dataset",
    data=filtered.to_csv(index=False).encode("utf-8"),
    file_name="filtered_happiness_2026.csv",
    mime="text/csv"
)

# -------------------------------------------------------------------------
# METHODOLOGY
# -------------------------------------------------------------------------

with st.expander("Methodology and Limitations"):
    st.markdown("""
    **Dataset:** World Happiness Update Report 2026 (Kaggle).

    **Method:** Cross-sectional descriptive and
    exploratory correlation analysis.

    **Data quality:** The original dataset contained
    147 records representing 130 unique country names.
    Seventeen countries had conflicting observations.
    Both observations for each affected country were
    excluded, leaving 113 countries.

    **Rankings:** Analysis ranks were recalculated
    for the retained sample. They are not official
    global rankings.

    **Interpretation:** Correlation does not establish
    causation. Indicators should not be interpreted
    as literal economic or health measurements without
    verifying their source definitions.

    **Coverage:** The dashboard represents only
    the validated, non-conflicting subset.
    """)

st.caption(
    "Global Happiness & Quality of Life Dashboard | "
    "Data Analytics Portfolio Project"
)
'''

(project / "app.py").write_text(
    app_code, encoding="utf-8"
)


# =============================================================================
# 3. PYTHON DEPENDENCIES
# =============================================================================

requirements = """streamlit
pandas
plotly
"""

(project / "requirements.txt").write_text(
    requirements, encoding="utf-8"
)


# =============================================================================
# 4. README DOCUMENTATION
# =============================================================================

readme = """# Global Happiness & Quality of Life Dashboard 2026

## Project Overview

An interactive data analytics dashboard examining
cross-country happiness scores and associated
quality-of-life indicators.

## Objectives

- Examine global happiness score distributions
- Compare country-level happiness scores
- Explore relationships between happiness and
  selected quality-of-life indicators
- Present findings through interactive visualisations

## Tools and Technologies

Python, Pandas, Plotly, Streamlit, Google Colab

## Dashboard Features

- Interactive world map
- Country rankings
- Happiness score distribution
- Pearson correlation heatmap
- Indicator scatter plots
- Country comparisons
- Interactive country filters
- Downloadable filtered data

## Dataset

Source: World Happiness Update Report 2026

https://www.kaggle.com/datasets/jummanalnahian/world-happiness-update-report-2026

The original dataset contains 147 records.

## Data Quality

The source contains 17 country names appearing twice
with conflicting scores and indicator values.

Both records for each affected country were excluded
from the dashboard dataset.

The retained dataset contains 113 countries.

Analysis ranks were recalculated within this subset
and must not be interpreted as official global ranks.

Excluded records are preserved separately for review.

## Methodology

Exploratory data analysis, descriptive statistics,
country comparisons and Pearson correlation analysis.

The dataset represents a single reporting year.
Correlations are associations, not causal evidence.

## Run Locally

Install dependencies:

pip install -r requirements.txt

Launch the application:

streamlit run app.py

## Project Structure

app.py
requirements.txt
happiness_2026_clean.csv
excluded_country_records.csv
README.md

## Limitations

- Conflicting source records affect country coverage.
- Original data provenance requires further verification.
- Country indicators require source-specific interpretation.
- Results cannot establish causation or temporal trends.
"""

(project / "README.md").write_text(
    readme, encoding="utf-8"
)


# =============================================================================
# 5. PACKAGE PROJECT FILES
# =============================================================================

zip_filename = "global_happiness_dashboard_2026.zip"

with zipfile.ZipFile(
    zip_filename, "w", zipfile.ZIP_DEFLATED
) as archive:
    for file in project.iterdir():
        archive.write(file, arcname=file.name)

print("\nPROJECT FILES")
for file in project.iterdir():
    print(file.name)

print("\nProject archive:", zip_filename)

files.download(zip_filename)
```
